```python
# 1. KHAI BÁO THƯ VIỆN
import evaluate
import numpy as np
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, TrainingArguments, Trainer
from peft import LoraConfig, get_peft_model, TaskType

print("1. Đang tải tập dữ liệu Tin tức Tài chính (Twitter Financial News)...")
# Dùng bộ dữ liệu tài chính siêu ổn định, không dính lỗi bảo mật
dataset = load_dataset("zeroshot/twitter-financial-news-sentiment")

# Bộ này xịn ở chỗ họ chia sẵn tập train và validation cho mình luôn
train_dataset = dataset["train"]
eval_dataset = dataset["validation"]

print("2. Đang tải Tokenizer và Base Model...")
model_name = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=3)

print("3. Đang mã hóa dữ liệu văn bản...")
def tokenize_function(examples):
    # Cột chứa văn bản của bộ này tên là 'text'
    return tokenizer(examples["text"], padding="max_length", truncation=True, max_length=128)

tokenized_train = train_dataset.map(tokenize_function, batched=True)
tokenized_eval = eval_dataset.map(tokenize_function, batched=True)

print("4. Cài đặt kỹ thuật LoRA (PEFT)...")
lora_config = LoraConfig(
    task_type=TaskType.SEQ_CLS,
    r=8,
    lora_alpha=16,
    lora_dropout=0.1,
    target_modules=["query", "value"]
)
model = get_peft_model(model, lora_config)

# Cực kỳ lưu ý dòng in ra này để lấy số liệu "Trainable Params" chép vào Word
model.print_trainable_parameters()

print("5. Thiết lập hàm tính điểm Macro F1...")
metric = evaluate.load("f1")
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    return metric.compute(predictions=predictions, references=labels, average="macro")

print("6. Cấu hình các thông số Huấn luyện (Training Arguments)...")
training_args = TrainingArguments(
    output_dir="./results",
    learning_rate=2e-4,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=5,
    eval_strategy="epoch",
    save_strategy="epoch",
    logging_steps=10
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_eval,
    compute_metrics=compute_metrics,
)

print("7. BẮT ĐẦU HUẤN LUYỆN! (Lần này chắc chắn mượt mà)")
trainer.train()
```

```python
# SCRIPT CHẠY THỰC NGHIỆM CHO BẢNG 2 (FINBERT)
from transformers import AutoTokenizer, AutoModelForSequenceClassification, TrainingArguments, Trainer
from datasets import load_dataset
import evaluate
import numpy as np

print("1. Tải dữ liệu và Mô hình FinBERT chuyên ngành...")
dataset = load_dataset("zeroshot/twitter-financial-news-sentiment")
model_name = "ProsusAI/finbert" # <--- Điểm khác biệt nằm ở đây
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=3)

# 2. Mã hóa dữ liệu
def tokenize_function(examples):
    return tokenizer(examples["text"], padding="max_length", truncation=True, max_length=128)

tokenized_train = dataset["train"].map(tokenize_function, batched=True)
tokenized_eval = dataset["validation"].map(tokenize_function, batched=True)

# 3. Thiết lập đo lường
metric = evaluate.load("f1")
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    return metric.compute(predictions=predictions, references=labels, average="macro")

# 4. Huấn luyện (Full Fine-tuning)
training_args = TrainingArguments(
    output_dir="./finbert_results",
    learning_rate=2e-5, # Tốc độ học thấp hơn cho Full Fine-tuning
    num_train_epochs=3,  # Chạy 3 epoch là đủ thấy sự vượt trội
    per_device_train_batch_size=16,
    eval_strategy="epoch",
    save_strategy="epoch"
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_eval,
    compute_metrics=compute_metrics,
)

print("5. Bắt đầu huấn luyện FinBERT...")
trainer.train()
```
